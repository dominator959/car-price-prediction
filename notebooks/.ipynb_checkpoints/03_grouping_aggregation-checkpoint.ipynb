# Phase 3: Exploratory Data Analysis (EDA)
- This notebook explores our cleaned Pakistani automotive marketplace dataset to uncover correlations, distribution patterns, and key pricing drivers before moving into model training.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the finalized clean data from the processed folder
df = pd.read_csv("../data/processed/03_final_cleaned_data.csv")

# Set global visualization settings
sns.set_theme(style="whitegrid")
%matplotlib inline

print(f"EDA Notebook Initialized. Data Shape: {df.shape}")
df.head()
```

# 1. Statistical Correlation
We use a Correlation Heatmap to determine which features have the strongest mathematical relationship with car price. This identifies the "heavy hitters" for our prediction model.

```python
# Select only numeric columns
numeric_cols = df.select_dtypes(include=[np.number])

# Calculate correlations
corr_matrix = numeric_cols.corr()

# Visualize Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap: What drives Price?')
plt.show()
```
