# df['price'].sample()
- What it does: Randomly selects and displays a specified number of samples (in this case, 10) from the price column.

# clean_and_scale_price()
- What it does: Takes a row value, looks for localized market terms (lac or crore), isolates the raw floating number, scales it by its matching mathematical factor ($100,000$ or $10,000,000$), and converts the outcome.
- Why we use it here: A dictionary map paired with an explicit custom function allows us to scale diverse local naming structures safely without encountering conversion errors.
- What to expect from it: Defining this logic executes silently, providing a bulletproof processing pipeline for our dataset.

```python
def clean_and_scale_price(x):
    # Ensure value is converted to a clean string format
    x = str(x).lower().strip()
    
    # Handle explicit currency prefixes or commas
    x = x.replace('pkr', '').replace(',', '').strip()
    
    # Mathematical multipliers based on local domain terminology
    if 'crore' in x:
        num_part = x.replace('crore', '').strip()
        return float(num_part) * 10000000
    elif 'lac' in x:
        num_part = x.replace('lacs', '').replace('lac', '').strip()
        return float(num_part) * 100000
        
    # Standard fallback for pure numeric text entries
    try:
        return float(x)
    except ValueError:
        return None
```

- This functions as a professional parsing tier. 
- By intercepting 'crore' before trying a direct float conversion, an entry like ' 1.5 crore' safely evaluates as $1.5 \times 10,000,000$, yielding a numerical value of 15000000.0.

# df['price'].apply() (With Multi-Tier Multipliers)
- What it does: Maps our robust scaling function across every cell in the price column, updating the table with correctly calculated values.

- Why we use it here: To safely run our multi-tier text conversions across the entire dataset without crashing on local denomination terms like "crore".

- What to expect from it: The column values will update into clean, uniform floats ready for statistical modeling.

```python
# Apply our new robust scaling function to the price column
df['price'] = df['price'].apply(clean_and_scale_price)

# Remove any rows that returned None due to unparseable text strings (e.g., "Call for price")
df.dropna(subset=['price'], inplace=True)
```

The processing loop now finishes without a single ValueError. By dropping unparseable rows at the end, we clear out edge cases like "Call for price" which cannot be converted into numbers. Your target variable is now fully standardized: '3.8 lacs' is stored as 380000.0 and '1.5 crore' is stored as 15000000.0.

# Cleaning the mileage Column
- What it does: Strips out the " km" text and the thousands-separator commas from every entry, then converts the remaining text into a pure integer.

- Why we use it here: Mileage is a critical "continuous feature" for predicting car value. Vehicles with higher mileage typically depreciate faster. To use this in a mathematical model, it must be a numeric type (int or float).

- What to expect from it: The column will change from strings like "100,000 km" to numbers like 100000.

```python
# 1. Remove 'km' and commas, then convert to numeric
# We use errors='coerce' to turn any non-numeric remains into NaN (missing values)
df['mileage'] = pd.to_numeric(
    df['mileage'].astype(str).str.replace('km', '', regex=False).str.replace(',', '', regex=False), 
    errors='coerce'
)

# 2. Briefly check the new distribution
df['mileage'].describe()
```


We used pd.to_numeric with errors='coerce'. This is a "safety first" approach—if there's a weird value like "Not Available" that we missed, Pandas will turn it into a NaN instead of crashing your entire notebook. The .describe() output will now show you the average distance traveled by cars in your dataset.

# Inspecting engine Capacity
- What it does: Randomly samples 5 entries from the engine column.

- Why we use it here: Typically, engine capacity is listed as "1300 cc" or "1.3L". We need to see the exact pattern to decide if we should store it as CC (cubic centimeters) or Liters.

- What to expect from it: A series of strings representing engine sizes.

```python
# Sample engine capacity to identify string patterns
df['engine'].sample(5, random_state=42)
```

This shows us two critical data problems we need to solve right now:

Formatting Noise: The valid values have a suffix of " cc". We need to strip this text and convert the column to numbers.

Missing Data (NaN): There are quite a few missing values here. Unlike the price column (our target variable, where dropping rows is mandatory), engine displacement is an independent feature. Dropping rows for every missing feature would shred our dataset size. Instead, we can use an advanced technique called Imputation—filling in the blanks using a smart statistical baseline like the median value of the column.

# Cleaning and Imputing the engine Column
- What it does: Strips the " cc" suffix, converts values to numbers, and fills missing entries.

- Why we use it here: Machine learning models require complete numerical data. We use Median Imputation because it is robust against outliers (like 5000cc engines) that would otherwise skew a simple average.

- Modern Standards: We avoid inplace=True by using explicit assignment (df['col'] = ...). This follows the latest Python performance guidelines and silences the FutureWarning.

```python
# 1. Clean the 'cc' text and convert to numeric format
df['engine'] = df['engine'].astype(str).str.replace('cc', '', regex=False).str.strip()
df['engine'] = pd.to_numeric(df['engine'], errors='coerce')

# 2. Calculate the median engine size from the valid rows
engine_median = df['engine'].median()
print(f"Calculated Median Engine Size: {engine_median} cc")

# 3. Impute (fill) the missing values (Modern Assignment Method)
df['engine'] = df['engine'].fillna(engine_median)

# 4. Verify that no missing values remain in this column
print("Remaining Missing Values in Engine:", df['engine'].isna().sum())
```

Your engine column is now a clean float64 type with a default value of 1300.0 standing in for any missing data.

# df['column'].value_counts()
- What it does: Scans a categorical column, identifies every unique text label present, and counts exactly how many rows are assigned to each label.

- Why we use it here: To check for class imbalances, find hidden missing values represented by text noise (like "Unknown" or "None"), and spot spelling inconsistencies.

- What to expect from it: A clean frequency distribution list sorted from the most common category to the rarest.

```python
# Check the unique categories inside transmission and fuel_type
print("--- Transmission Value Counts ---")
print(df['transmission'].value_counts(dropna=False))

print("\n--- Fuel Type Value Counts ---")
print(df['fuel_type'].value_counts(dropna=False))
```

The Breakdown of What’s Wrong:
transmission has 708 NaN (Missing) values: Over half of your data entries are missing transmission tags!

fuel_type contains "4 Stroke" and "2 Stroke": Wait a minute... cars do not have 2-stroke engines! 2-stroke and 4-stroke classifications belong to motorcycles and rickshaws. This means your dataset contains scraped scrap or noise from mixed automotive categories that will completely confuse a car-price prediction model.

Let's fix the fuel_type contamination first by filtering our dataset down to only actual car fuels: Petrol, Hybrid, Electric, Diesel, LPG, PHEV, and CNG.

# Filtering Out Bike/Rickshaw Contamination from fuel_type
- What it does: Drops any rows where the fuel type is recorded as "4 Stroke" or "2 Stroke", leaving behind only valid car-based engine fuels.

- Why we use it here: Data quality is paramount. A car price model cannot learn from motorcycle engine profiles. Removing these rows keeps our dataset highly targeted and accurate.

- What to expect from it: A reduced but clean DataFrame focused solely on cars.

```python
# List of valid automotive fuel types
valid_car_fuels = ['Petrol', 'Hybrid', 'Electric', 'Diesel', 'LPG', 'PHEV', 'CNG']

# Filter the dataframe to only keep valid car fuels
df = df[df['fuel_type'].isin(valid_car_fuels)]

# Print new shape to see how many valid rows remain
print("Remaining rows after filtering:", df.shape[0])
```

This step sanitizes our data lake. By removing the 2-stroke and 4-stroke entries, we prune out the non-car anomalies entirely, ensuring our feature set strictly represents the vehicle domain we are trying to predict.

# Re-evaluating Transmission for the Cleaned Dataset

- What to expect from it: A new frequency list showing the state of our remaining 619 car records.

```python
# Check transmission counts for our 619 remaining cars
print("--- Updated Transmission Counts ---")
print(df['transmission'].value_counts(dropna=False))
```

# Strategic Imputation for transmission
- What it does: Fills any remaining NaN values in the transmission column with the Mode (the most frequent value, likely "Automatic").

- Why we use it here: Since transmission is a category (text), we can't use a "mean." We use the most common value to fill gaps so our model doesn't crash, assuming the "average" car in this specific filtered set follows the majority trend.

- What to expect from it: A fully populated transmission column with no missing values.

```python
# 1. Find the most common transmission (Mode)
trans_mode = df['transmission'].mode()[0]
print(f"The most common transmission is: {trans_mode}")

# 2. Fill the missing values with the mode
df['transmission'] = df['transmission'].fillna(trans_mode)

# 3. Final verification of all columns
df.isna().sum()
```

This completes our "Missing Value Treatment" phase. By the end of these cells, your df.isna().sum() should show 0 for every single column. This is the "Gold Standard" state for data—no holes, no noise, just clean numbers and categories.
