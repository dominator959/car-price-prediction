# Importing Required Libraries

- pandas is the main library for loading and working with tabular data
- We import it with the standard alias pd
- This is the first step in every data science notebook

```python
import pandas as pd
```

# Loading the Raw Dataset

- We load the data from the raw/ folder
- Always use a variable for the file path (good professional habit)
- This step shows us how many car listings we have

```python
# Load the raw PakWheels dataset
raw_data_path = '../data/raw/pakwheels.csv'   # Update filename if different
df = pd.read_csv(raw_data_path)

print(f"Shape: {df.shape[0]:,} rows and {df.shape[1]} columns")
```

# Checking Column Names

- We inspect the exact column names to understand what information is available
- This helps us plan which features are useful for predicting car price
- Common columns in such datasets: Make, Model, Year, Price, KM, etc.

```python
# Display all column names
print("Column Names:")
for col in df.columns:
    print(f"- {col}")
```
