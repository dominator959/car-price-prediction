# Importing Required Libraries

- pandas is the main library for loading and working with tabular data
- We import it with the standard alias pd
- This is the first step in every data science notebook

```python
import pandas as pd
```

# Loading the Raw Dataset

- We load the data from the raw/ folder
- Always use a variable for the file path (good professional habit)
- This step shows us how many car listings we have

```python
# Load the raw PakWheels dataset
raw_data_path = '../data/raw/pakwheels.csv'   # Update filename if different
df = pd.read_csv(raw_data_path)

print(f"Shape: {df.shape[0]:,} rows and {df.shape[1]} columns")
```

# Checking Column Names

- We inspect the exact column names to understand what information is available
- This helps us plan which features are useful for predicting car price
- Common columns in such datasets: Make, Model, Year, Price, KM, etc.

```python
# Display all column names
print("Column Names:")
for col in df.columns:
    print(f"- {col}")
```

# First Look at the Data

- df.head() shows the first 5 rows
- This helps us understand what each column actually contains
- We can see real examples of car listings (price, year, mileage, etc.)

```python
# Display first 5 rows
df.head()
```

# Better View of First Few Rows

- We use df.head(5).T to transpose the data
- This makes it easier to read all columns clearly for each car
- This view helps us understand the structure better

```python
# Better view - transpose first 5 rows
df.head(5).T
```

# Basic Information About Dataset

- df.info() shows data types and missing values
- This helps us identify which columns need cleaning
- We expect many object (text) columns that will need conversion

```python
# Get basic information about the dataset
df.info()
```

# Checking Missing Values

- df.isnull().sum() shows exactly how many missing values are in each column
- This is crucial for planning our data cleaning strategy
- We expect high missing values in make, model_slug, engine, etc.

```python
# Count missing values in each column
missing_values = df.isnull().sum()
print(missing_values)
```

# Statistical Summary of Numerical Columns

- df.describe() gives us count, mean, min, max for numerical columns
- This helps us understand the range of year and spot any unrealistic values early
- We will focus on year for now as it is the only useful numeric column currently

```python
# Get statistical summary
df.describe()
```

# Data Loading Summary

- We have successfully loaded and inspected the raw dataset
- Key findings: 1,394 total rows, many missing values, price and mileage stored as text
- This completes the initial exploration of the PakWheels dataset

```python
# Final summary of data loading
print("=== 01_data_loading.ipynb Summary ===")
print(f"Total rows loaded     : {len(df):,}")
print(f"Rows with price data  : {df['price'].notnull().sum():,}")
print(f"Year range            : {df['year'].min()} - {df['year'].max()}")
print(f"Columns with missing values: {df.isnull().sum()[df.isnull().sum() > 0].count()}")
print("\n Data loading completed!")
```
